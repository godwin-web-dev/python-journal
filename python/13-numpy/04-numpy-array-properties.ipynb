```python
#Numpy array properties
import numpy as np

np_array=np.array([[1,2,3],
                  [4,5,6]])

# shape of the array it gives info about no of rows and columns
print("shape of array is \n",np_array.shape)

# size of the array it gives info about no. of elements in array
print("size of array is \n",np_array.size)

# dimension of the array it gives info about depth of the array
print("dimension of array is \n",np_array.ndim)

# to print data type of an array
print("datatype of array is \n",np_array.dtype)
```

```python
#note if any one of the values are changed from the int to float
np_array=np.array([[1,2],
                   [3,4]])
print("datatype of the array is ",np_array.dtype)
np_new_array=np.array([[1,2],
                       [3,4.5]])
print("data type after changing the 4 to 4.5 \n",np_new_array.dtype)
```
