```python
import numpy as np
np_array=np.array([1,2,3,4])
np_random_array=np.random.randint(0,100,size=(4,4))
print("random array is \n",np_random_array)
np_zeroes=np.zeros((3,3))
print("zeroes array is \n",np_zeroes)

# to save this arrays 
np.save('normal-array.npy',np_array)
np.save('random-array.npy',np_random_array)

# to load the saved array 
np.load('normal-array.npy')
np.load('random-array.npy')
```

```python
import matplotlib.pyplot as plt
load_numpy_img=np.load('numpy-logo.npy')

plt.imshow(load_numpy_img)
plt.title('white mode numpy image')
plt.show()

# darkmode numpy image is 
dark_mode_img=1-load_numpy_img
plt.imshow(dark_mode_img)
plt.title('numpy dark mode image')
plt.show()
```
