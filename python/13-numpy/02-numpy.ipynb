```python
import numpy as np
one_d_array=np.array([1,2,3])
print("one dimensional array is ",one_d_array)
two_d_array=np.array([[1,2,3],[3,4,5]])
print("two dimensional array is ",two_d_array)
```

```python
##To print numbers in np array
numpy_range_array=np.arange(1,100)
print("numpy array from 1 to 100 is ",numpy_range_array)
```

```python
#Different ways to create an array from the scratch 
import random 
# to create an array with zeroes
np_zeroes=np.zeros((3,3))
print("zeroes array is as follow \n",np_zeroes)

# to create an array with ones 
np_ones_array=np.ones((3,3))
print("ones array is as follow \n",np_ones_array)

# to create an array with constant number
np_constant_array=np.full((3,3),8)
print("constant array is as follow \n",np_constant_array)

# an array of random numbers
# syntax => start,stop,dimension of array either 4 X 4 like that
np_random_array=np.random.randint(0,20,(4,4))
print("random array is as follow \n",np_random_array)

# an array of sequence
# arange generates a sequnece of number that requires the start range  and the end range and u can control the sequence using the step 
np_sequence_array=np.arange(1,100,3)
print("sequence array is as follow \n",np_sequence_array)
```
