```python
import pandas as pd
import numpy as np
# creating a series 
pd_series=pd.Series(np.arange(1,100))
# to print the datatype of the pdseries
print("pd series datatype",pd_series.dtypes)
print(pd_series)

# to print the type of pdseries
type(pd_series)
# pandas.core.series.Series
```

```python
my_custom_dataframe=pd.DataFrame(np.random.rand(300,5),index=np.arange(300))
print("my dataframe ")
my_custom_dataframe


# dataframe.T transpose the row-> column and column-> row
my_custom_dataframe.T
```

```python
my_custom_dataframe.head()
```

```python
# to find the datatype of dataframe
my_custom_dataframe.dtypes
```

```python
my_custom_dataframe[0][0]="godwin"
my_custom_dataframe.dtypes

# if u convert the any one row of the column entire datatype of the column will be changed 
```

```python
# tosort the dataframe
# sort ascending with horizontal axis
my_custom_dataframe.sort_index(axis=0,ascending=True)

# sort descending with vertical axis
my_custom_dataframe.sort_index(axis=1,ascending=False).head()
```

```python
copying_dataframe=my_custom_dataframe
copying_dataframe
```

```python
copying_dataframe[0][0]="roshan"
# note if u try to copy the dataframe like this internally both the dataframes are pointing to the same memory location so changes done in the copied dataframe will also be reflected in the original dataframe
# copying_dataframe 
my_custom_dataframe
```

```python
# copying the dataframe without modifying the original data
new_copy2=my_custom_dataframe.copy()
print(new_copy2)

new_copy2[0][0]="roshan"
print("dataframe after modifying ")
print(new_copy2)

print("original dataframe ")
print(my_custom_dataframe)
```

```python
# To avoid issues with shallow copying and also the SettingWithCopyWarning, use the command below.
# .loc stands for "location" and is used for label-based indexing to safely assign values.
# This is the recommended way for modifying DataFrame values.
copying_dataframe.loc[0,0]="god"
copying_dataframe
```

```python
# drops the entire column 
copying_dataframe.drop(0,axis=1)
```
